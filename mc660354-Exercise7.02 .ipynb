```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
data = pd.read_csv("northern_surface_temperature.csv", index_col=['Year'])
data = data.transpose()
```

```python
heat_colormap = sns.diverging_palette(240, 15, s=99, as_cmap=True)
```

```python
plt.figure(dpi=200)
sns.heatmap(data.iloc[:, ::5], cmap=heat_colormap, center=0)
plt.title("Temperature Changes from 1880 to 2015 (base period 1951-1980)")
plt.savefig('temperature_change.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 计算每年的平均温度
yearly_temp = data.mean()

plt.figure(dpi=150)
sns.lineplot(x=yearly_temp.index, y=yearly_temp.values)
plt.title("Average Temperature Anomalies Over Years (Northern Hemisphere)")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```
